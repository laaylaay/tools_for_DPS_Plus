```python
import os
import json

import torch
from transformers import AutoProcessor, Qwen2_5_VLForConditionalGeneration

from weak2strong_tool import weak2strong_tool
```

Weak2Strong Demo

```python
model_name = "Qwen/Qwen2.5-VL-3B-Instruct"
dataset = "Self-Gen"
method = "weak2strong_tool"
label = "Abyssinian"
attack_text = "Somali"
mask_percentile = 80
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
local_model = Qwen2_5_VLForConditionalGeneration.from_pretrained(
    model_name, torch_dtype=torch.bfloat16, attn_implementation="eager"
).eval().to(device)
local_processor = AutoProcessor.from_pretrained(model_name, trust_remote_code=True)
```

```python
files = os.listdir("./examples/")
```

```python
output_path = f"./{model_name}_{dataset}_{method}.jsonl"
for i in files:
    image_path = "./examples/" + i
    result = weak2strong_tool(
        local_model=local_model,
        local_processor=local_processor,
        image_path=image_path,
        label=label,
        attack_text=attack_text,
        mask_percentile=mask_percentile,
    )
    result["image_path"] = image_path
    with open(output_path, "a") as jsonl_file:
        jsonl_file.write(json.dumps(result, ensure_ascii=False) + "\n")
```
